# 📊 01. Exploratory Data Analysis (EDA) - Customer Churn Prediction
### TY BSc Data Science Final-Year Project
**Objective:** Analyze the IBM Telco customer dataset to identify customer attrition patterns, demographic distributions, service preferences, and billing characteristics driving churn.

```python
import os
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configure visual style
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
print("Libraries loaded successfully.")
```

## 1. Load Dataset & Initial Inspection
We load the raw dataset located at `../data/raw/customer_churn.csv`.

```python
raw_data_path = os.path.join("..", "data", "raw", "customer_churn.csv")
df = pd.read_csv(raw_data_path)
print(f"Dataset Dimensions: {df.shape[0]} rows x {df.shape[1]} columns")
df.head()
```

## 2. Data Types & Missing Value Audit

```python
df.info()
print("\nMissing values count:")
print(df.isnull().sum())
```

Notice that `TotalCharges` is stored as an `object` type because blank spaces `' '` exist for new accounts where `tenure == 0`.

```python
# Check blank string values in TotalCharges
blank_charges = df[df['TotalCharges'].str.strip() == '']
print(f"Rows with blank TotalCharges: {len(blank_charges)}")
blank_charges[['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges']]
```

## 3. Target Class Distribution (Churn Analysis)
Let's examine the baseline churn rate to assess class imbalance.

```python
churn_counts = df['Churn'].value_counts()
churn_pct = df['Churn'].value_counts(normalize=True) * 100

print("Churn Counts:\n", churn_counts)
print("\nChurn Percentage:\n", churn_pct)

plt.figure(figsize=(6, 5))
plt.pie(churn_counts, labels=['Retained (No)', 'Churned (Yes)'], autopct='%1.1f%%',
        colors=['#2b5c8f', '#d9534f'], explode=(0, 0.08), startangle=140, shadow=True)
plt.title("Overall Customer Churn Distribution", fontsize=14, fontweight='bold')
plt.show()
```

**Observation:** Approximately 26.5% of customers churn, while 73.5% are retained. This moderate imbalance means standard accuracy alone is insufficient; Recall and F1-score are critical evaluation metrics.

## 4. Key Churn Drivers: Contract Type & Tenure

```python
plt.figure(figsize=(9, 5))
sns.countplot(data=df, x='Contract', hue='Churn', palette=['#2b5c8f', '#d9534f'])
plt.title("Customer Churn by Contract Type", fontsize=14, fontweight='bold')
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")
plt.show()
```

**Insight:** Customers on Month-to-month contracts exhibit an overwhelmingly higher churn rate compared to those on 1-year or 2-year commitments.

```python
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='tenure', hue='Churn', kde=True, multiple='stack', palette=['#2b5c8f', '#d9534f'])
plt.title("Tenure Distribution by Churn Status", fontsize=14, fontweight='bold')
plt.xlabel("Tenure (Months)")
plt.ylabel("Customer Count")
plt.show()
```

## 5. Billing Charges & Payment Method

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x='PaymentMethod', y='MonthlyCharges', hue='Churn', palette=['#2b5c8f', '#d9534f'])
plt.xticks(rotation=20)
plt.title("Monthly Charges vs Payment Method by Churn Status", fontsize=14, fontweight='bold')
plt.show()
```

## 6. Summary of Key EDA Findings
1. **Contract Type:** Month-to-month contracts have the highest attrition risk (>40%).
2. **Tenure:** First 12 months are the most vulnerable customer window.
3. **Monthly Charges:** Churners have substantially higher average monthly charges ($74 vs $61).
4. **Payment Method:** Electronic Check users churn at significantly higher rates than auto-pay users.
